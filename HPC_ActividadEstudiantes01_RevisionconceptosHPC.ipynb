## **Revisión Exhaustiva de Conceptos y Tendencias en HPC**

### **Objetivo**

El propósito de esta actividad es que los estudiantes realicen una revisión exhaustiva de los conceptos fundamentales, principios, herramientas, software, hardware y tendencias actuales y futuras en el campo de la Computación de Alto Desempeño (HPC). Además, deberán explorar ejemplos prácticos donde estas tecnologías sean aplicadas en la actualidad o tengan un alto potencial en el futuro.

### **Indicaciones Generales**

1. **Formato y Entrega:**
   - La actividad puede realizarse en parejas.
   - El informe debe entregarse en formato jupyter Notebook o PDF.
   - Debe incluir los elementos principales de un trabajo escrito: Introducción, objetivos, cuerpo del trabajo, referencias bibliográficas debidamente indicadas en el cuerpo del trabajo, bibliografía.
   - Fecha de entrega: viernes marzo 07 antes de las 4:00 pm en el espacio de la UVirtual creado para tal fin.

2. **Fuentes de Información:**
   - El material base recomendado es la documentación disponible en el enlace: [https://hpc.llnl.gov/documentation/tutorials/introduction-parallel-computing-tutorial](https://hpc.llnl.gov/documentation/tutorials/introduction-parallel-computing-tutorial).
   - Se espera que los estudiantes consulten fuentes adicionales como artículos científicos, blogs especializados, conferencias y documentación de fabricantes de hardware y software HPC.

3. **Estructura del Informe:**
   - **Introducción:** Breve presentación del tema, importancia del HPC y objetivos del documento.
   - **Conceptos Fundamentales de HPC:** Definición y principios básicos.
   - **Software y Hardware en HPC:**
     - Descripción de arquitecturas modernas (CPU, GPU, FPGA, computación cuántica y otras emergentes).
     - Sistemas operativos y software clave en HPC.
   - **Tendencias Actuales y Futuras:**
     - Impacto de la inteligencia artificial en HPC.
     - Computación cuántica y su potencial.
     - Avances en redes de interconexión y almacenamiento para HPC.
   - **Ejemplos Prácticos:**
     - Casos de uso en la industria, ciencia e ingeniería.
     - Descripción breve de aplicaciones que dependen de HPC.
   - **Conclusiones y Reflexión:**
     - Perspectivas futuras y desafíos en HPC.
   - **Bibliografía:** Listado de fuentes utilizadas en formato APA o IEEE. No es suficiente con colocar un listado de fuentes bibliográficas. Cada ítem debe estar debidamente referenciado en el cuerpo del trabajo. 


### **Puntos a Desarrollar**

1. **Conceptos Fundamentales**
   - Explicar los principios de paralelismo en computación.
   - Diferencias entre paralelismo a nivel de instrucciones, datos y tareas.

2. **Arquitecturas y Software en HPC**
   - Describir las arquitecturas utilizadas en HPC y sus ventajas.
   - Herramientas y frameworks populares en el desarrollo de aplicaciones HPC.

3. **Tendencias Actuales y Futuras**
   - Breve análisis sobre el impacto de la computación cuántica en HPC.
   - ¿Cómo está evolucionando la supercomputación en la actualidad?
   - Rol de la inteligencia artificial en la optimización de sistemas HPC.

4. **Ejemplos Prácticos**
   - Presentar al menos dos aplicaciones reales donde el HPC sea clave.
   - Mencionar qué tipo de hardware y software se utiliza en estos casos.

5. **Análisis Crítico y Reflexión Final**
   - Opinión fundamentada sobre las oportunidades y desafíos del HPC.
   - Consideraciones sobre el futuro de la computación paralela.
