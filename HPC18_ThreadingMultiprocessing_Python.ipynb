## **Introducción**

La programación concurrente juega un papel fundamental en la optimización del rendimiento y la eficiencia en las aplicaciones modernas. En Python, existen dos enfoques principales para lograr la concurrencia: **Threading** y **Multiprocessing**. A lo largo de este texto, exploraremos las diferencias entre ambos enfoques, sus ventajas y casos de uso ideales. Además, realizaremos una demostración práctica mediante la descarga de imágenes, evaluando el rendimiento de cada técnica para ilustrar cuándo es más conveniente utilizar uno u otro enfoque.

## **Concurrencia vs. Paralelismo**

### **Concurrencia (Threading)**

La **concurrencia** es la capacidad de un programa para manejar múltiples tareas de forma simultánea. No implica la ejecución simultánea real en diferentes núcleos de CPU, sino que permite la superposición de tareas mediante el uso de "cambio de contexto" (context switching). Así, las tareas parecen ejecutarse al mismo tiempo al alternar rápidamente entre ellas. Esta técnica se emplea en un único núcleo de CPU, maximizando su uso al intercambiar entre las tareas en intervalos cortos de tiempo. 

Un ejemplo clásico de uso de concurrencia es la descarga de archivos desde la web o la lectura y escritura de datos, donde el programa puede aprovechar los tiempos de espera de una operación de entrada/salida (I/O) para realizar otra tarea, sin detener completamente la ejecución.

### **Paralelismo (Multiprocessing)**

El **paralelismo**, por otro lado, se refiere a la ejecución simultánea real de tareas en múltiples núcleos de CPU. Aquí, las tareas son independientes y se dividen entre varios procesos que se ejecutan en paralelo, lo que permite acelerar significativamente la ejecución. El paralelismo es ideal para tareas que requieren un uso intensivo de CPU, como el procesamiento de grandes volúmenes de datos o cálculos matemáticos complejos, donde cada núcleo puede encargarse de una parte de la tarea.

Mientras que la concurrencia ofrece la apariencia de simultaneidad dentro de un solo núcleo, el paralelismo utiliza múltiples núcleos para lograr una verdadera simultaneidad.

### **Casos de Uso**

- **Concurrencia (Threading):** Es ideal para tareas **I/O-bound**, donde el programa debe esperar por operaciones externas, como leer archivos, hacer solicitudes a APIs o interactuar con bases de datos. Utilizar hilos permite solapar estas operaciones de espera con otras tareas, mejorando la eficiencia general.
  
  Ejemplo: El scraping web es un caso donde la concurrencia es útil, ya que permite realizar múltiples solicitudes simultáneamente para recopilar datos de varias páginas web, mejorando el tiempo de respuesta.
  
- **Paralelismo (Multiprocessing):** Es más adecuado para tareas **CPU-bound**, donde el tiempo de ejecución depende de cálculos intensivos. Al dividir la carga de trabajo entre múltiples núcleos de CPU, se puede acelerar significativamente el procesamiento de datos.

  Ejemplo: El procesamiento de grandes volúmenes de datos, donde cada núcleo de la CPU se encarga de una porción de los datos, es un caso típico de paralelismo.

### **Desventajas**

Ambos enfoques presentan retos y sobrecargas asociadas.

- **Concurrencia (Threading):** 
  - **Sobrecarga de sincronización:** Coordinar los recursos compartidos entre hilos puede ser complejo, y los errores de sincronización pueden generar condiciones de carrera o bloqueos.
  - **Dificultad para depurar:** Los programas concurrentes son más difíciles de depurar debido al comportamiento no determinista generado por la intercalación de los hilos.
  
- **Paralelismo (Multiprocessing):**
  - **Sobrecarga de comunicación:** La coordinación entre procesos requiere mecanismos explícitos como comunicación entre procesos (IPC) o memoria compartida, lo que puede afectar el rendimiento.
  - **Desbalance de carga:** En entornos paralelos, es crucial que las tareas estén bien distribuidas entre los procesos; de lo contrario, algunos procesos pueden quedar subutilizados.

### **Threading en Python**

El **Threading** en Python permite la ejecución concurrente de tareas utilizando múltiples hilos dentro de un único proceso. Estos hilos comparten el mismo espacio de memoria, pero debido al **[Global Interpreter Lock (GIL)](https://realpython.com/python-gil/)**, la ejecución de código en múltiples hilos no puede aprovechar completamente múltiples núcleos de CPU. Por esta razón, el threading es más apropiado para tareas **I/O-bound**, donde los hilos pueden esperar sin bloquear todo el proceso.

**Ejemplo de Threading**

```python
# importar el modulo threading
import threading

# definición de la función que se ejecuta con threadings
def funcion_hilo(nombre):
    print(f"Hola desde el hilo {nombre}")

# creación de los hilos (threads) y ejecución de la función
hilo = threading.Thread(target=funcion_hilo, args=("Hilo 1",))
hilo.start()
hilo.join()
```

### **Multiprocessing en Python**

El **Multiprocessing** permite un verdadero paralelismo al utilizar múltiples procesos, cada uno con su propio espacio de memoria. Esto hace que el multiprocessing sea ideal para tareas **CPU-bound** que pueden distribuirse entre varios núcleos.

**Ejemplo de Multiprocessing**

```python
# importar el modulo multiprocessing
import multiprocessing

# definición de la función que se ejecuta en paralelo
def funcion_proceso(nombre):
    print(f"Hola desde el proceso {nombre}")

# Inicio del procesamiento paralelo.
proceso = multiprocessing.Process(target=funcion_proceso, args=("Proceso 1",))
proceso.start() # inicializa el proceso paralelo
proceso.join()  # finaliza el proceso paralelo
```

### **Comparativa: Threading vs. Multiprocessing**

Realizaremos una comparativa de rendimiento mediante una tarea práctica de descarga de imágenes usando ambos enfoques.

#### **Código Base para la Comparativa**

```python
import timeit
import requests
import threading
import multiprocessing
```

Crear una lista para almacenar las imágenes descargadas desde la URL del servicio `picsum.photos`

```python
image_urls = []
for i in range(100,500):
    image_urls.append(f'https://picsum.photos/{i}')

print(image_urls)
```

La función `descargar_imagen` se define para gestionar la descarga de imágenes. Toma como parámetro una URL de imagen (`image_url`) y utiliza la biblioteca `requests` para obtener la imagen desde la URL. Si la descarga es exitosa (con un código de estado 200), la imagen se guarda en el directorio `images` con un nombre de archivo derivado de la URL. Si ocurre un error durante la descarga, se imprime un mensaje de error.

```python
def descargar_imagen(url):
    respuesta = requests.get(url)
    if respuesta.status_code == 200:
        with open(f"images/{url.split('/')[-1]}.png", 'wb') as f:
            f.write(respuesta.content)
    else:
        print(f"Error al descargar la imagen {url}")
```

### **Ejecución Normal (Secuencial)**

```python
def ejecucion_normal():
    inicio = timeit.default_timer()
    for url in image_urls:
        descargar_imagen(url)
    fin = timeit.default_timer()
    print(f'Tiempo de Ejecución Normal: {fin-inicio}')
    
ejecucion_normal()
```

#### **Ejecución con Threading**

```python
def descarga_threading():
    inicio = timeit.default_timer()
    threads = []
    for url in image_urls:
        t = threading.Thread(target=descargar_imagen, args=(url,))
        threads.append(t)
    for hilo in threads:
        hilo.start()
    for hilo in threads:
        hilo.join()
    fin = timeit.default_timer()
    print(f'Tiempo de Ejecución con Threading: {fin-inicio}')

descarga_threading()
```

### **Ejecución con Multiprocessing**

La función `descarga_multiprocessing` en Python utiliza el módulo de multiprocesamiento para ejecutar en paralelo la descarga de imágenes desde una lista de URLs, creando un proceso independiente para cada imagen. Los procesos se gestionan con los métodos `start()` y `join()`, y se mide el tiempo de ejecución para evaluar el rendimiento. La función `descargar_imagen` debe estar definida para que la descarga funcione correctamente.

```python
def descarga_multiprocessing():
    inicio = timeit.default_timer()
    procesos = []
    for url in image_urls:
        proceso = multiprocessing.Process(target=descargar_imagen, args=(url,))
        procesos.append(proceso)
    for proceso in procesos:
        proceso.start()
    for proceso in procesos:
        proceso.join()
    fin = timeit.default_timer()
    print(f'Tiempo de Ejecución con Multiprocessing: {fin-inicio}')

descarga_multiprocessing()
```

### **Conclusión**

**Threading** y **Multiprocessing** en Python son técnicas poderosas que permiten la ejecución concurrente y paralela de tareas, respectivamente. El threading es más adecuado para tareas I/O-bound, mientras que el multiprocessing es ideal para tareas CPU-bound. Aunque threading tiene limitaciones debido al GIL, sigue siendo útil para aplicaciones que requieren concurrencia sin un alto uso de CPU. Por su parte, multiprocessing permite un verdadero paralelismo, aunque a costa de una mayor sobrecarga en recursos del sistema.

El uso adecuado de cada enfoque depende del tipo de tarea a ejecutar y los recursos disponibles. Es fundamental comprender las diferencias y limitaciones para lograr un rendimiento óptimo.

## **Comparación directa entre Python y C**



| Característica               | Threading (Python)            | Multiprocessing (Python)    | OpenMP (C)                   | MPI (C)                      |
|------------------------------|-------------------------------|-----------------------------|------------------------------|------------------------------|
| **Modelo de paralelismo**     | Concurrencia simulada (GIL)    | Paralelismo real (procesos)  | Paralelismo real (hilos)      | Paralelismo distribuido       |
| **Uso de múltiples núcleos**  | Limitado por el GIL            | Completo (sin GIL)           | Completo                     | Completo                     |
| **Memoria compartida**        | Sí (entre hilos)               | No (procesos separados)      | Sí (entre hilos)              | No (memoria distribuida)      |
| **Comunicación**              | Fácil entre hilos (mismo proceso) | Requiere IPC (más costoso)   | Natural (memoria compartida)  | Paso de mensajes (más costoso)|
| **Escalabilidad**             | Limitado a una máquina         | Limitado a una máquina       | Escalable en memoria compartida | Escalable en clústeres       |
| **Simplicidad de uso**        | Sencillo, pero limitado        | Relativamente simple         | Sencillo con directivas       | Complejo, control manual      |
| **Casos de uso ideal**        | Tareas I/O-bound               | Tareas CPU-bound en una máquina | Tareas CPU-bound en memoria compartida | HPC, simulaciones a gran escala |


## **¿Cuándo usar uno u otro?**

**Usar Threading en Python**:
- Cuando las tareas son **I/O-bound** (lectura/escritura de archivos, operaciones de red).
- Cuando el paralelismo real no es necesario, y la concurrencia es suficiente.
- Cuando se busca una solución sencilla y fácil de implementar.

**Usar Multiprocessing en Python**:
- Cuando las tareas son **CPU-bound** y necesitan aprovechar múltiples núcleos.
- Cuando se busca paralelismo real dentro de una sola máquina sin las limitaciones del GIL.
- Cuando se pueden tolerar los costos de memoria y comunicación entre procesos.

**Usar OpenMP en C**:
- Cuando las tareas son intensivas en CPU y requieren **paralelismo real** en múltiples núcleos.
- Cuando el sistema tiene una arquitectura de **memoria compartida**.
- Cuando se desea una paralelización sencilla de bucles o secciones críticas del código.

**Usar MPI en C**:
- Cuando se necesita **paralelismo distribuido** en sistemas de memoria distribuida (clústeres, supercomputadoras).
- Para simulaciones y aplicaciones de computación de alto rendimiento (HPC).
- Cuando la aplicación requiere escalar a miles de nodos y puede tolerar la complejidad adicional de la programación y la comunicación interproceso.
